```python
clean_date_strings = pd.to_numeric(df['registration_init_time'], errors='coerce') \
                           .astype('Int64') \
                           .astype(str) \
                           .str.replace('<NA>', '')  # Strip pandas' missing integer string representation
    
# Parse the standardized YYYYMMD string values into a standard datetime object
reg_dates = pd.to_datetime(clean_date_strings, format='%Y%m%d', errors='coerce')
snapshot_date = pd.Timestamp('2017-02-28')  
# Compute the exact time active span in days (missing accounts remain as NaN for now)
df['time_active'] = (snapshot_date - reg_dates).dt.days.astype('float32')
member_numerical_categoricals = ['city', 'bd', 'registered_via', 'registration_init_time', 'time_active']

print("Imputing -1 for numerical and categorical profile attributes...")
for col in member_numerical_categoricals:
    if col in df.columns:
        df[col] = df[col].fillna(-1)
        
        # Downcast primitives where appropriate to save memory footprint
        if col in ['city', 'registered_via']:
            df[col] = df[col].astype('int8')
        elif col in ['bd']:
            df[col] = df[col].astype('int16')
        elif col in ['registration_init_time', 'time_active']:
            df[col] = df[col].astype('float32') # Keeping float32 allows downstreams to know it's a numeric continuum

print("Imputing 'not_spec' for demographic gender gaps...")
if 'gender' in df.columns:
    df['gender'] = df['gender'].fillna('not_spec').astype(str)
```

```python
print("--- STEP 1: RESOLVING THE SINGLE-TRANSACTION MODE GAPS ---")

# For users with exactly 1 transaction, copy their latest data to establish their historical baseline
mode_fallbacks = {
    'hist_ex_latest_mode_payment_plan_days': 'hist_latest_payment_plan_days',
    'hist_ex_latest_mode_plan_list_price': 'hist_latest_plan_list_price',
    'hist_ex_latest_mode_actual_amount_paid': 'hist_latest_actual_amount_paid',
    'hist_ex_latest_mode_is_auto_renew': 'hist_latest_is_auto_renew'
}

for mode_col, latest_col in mode_fallbacks.items():
    if mode_col in df.columns and latest_col in df.columns:
        initial_nan = df[mode_col].isna().sum()
        df[mode_col] = df[mode_col].fillna(df[latest_col])
        post_nan = df[mode_col].isna().sum()
        print(f"-> Mode {mode_col:<40} | Resolved {initial_nan - post_nan:,} single-tx records. Remaining NaNs: {post_nan:,}")


print("\n--- STEP 2: CONFLICT-FREE IMPUTATIONS FOR ZERO-TRANSACTION POOL (2,527 Users) ---")

# 1. Dedicated Sentinel Fills (Prevents overlapping with valid 0 or stable flags)
explicit_sentinel_fills = {
    'recent_autorenew_intent_trend': -9,       # -9 = Missing history (0 = Stable sequence)
    'latest_auto_renew_dropped': -9,           # -9 = Missing history (0 = No drop occurred)
    'hist_latest_is_auto_renew': -1,           # -1 = No billing records (0 = Manual renewer)
    'hist_ex_latest_mode_is_auto_renew': -1,   # -1 = No billing records (0 = Manual mode)
    'hist_latest_is_cancel': -1,               # -1 = No billing records (0 = Active/No cancel)
    'payment_method_mode': '-1',               # '-1' = Unknown payment channel string
    'hist_recency_days': -999,                 # Out-of-bounds timeline flags
    'hist_expiration_recency_days': -999,      # Out-of-bounds timeline flags
    'replenishment_consistency_variance': -1.0 # Standard native missing token
}

# 2. Pure Numeric Zero Fills (Where 0 is the literal absolute truth of having no transactions)
absolute_zero_fills = [
    'hist_transaction_count', 'hist_total_cancels', 'hist_payment_method_change', 
    'hist_tenure_days', 'hist_cancel_rate', 'payment_toggle_state_flips', 
    'payment_elasticity_vector', 'payment_discount_amount', 'payment_discount_ratio', 
    'is_promo_tier_subscriber', 'hist_ex_latest_mean_payment_plan_days', 
    'hist_ex_latest_mean_actual_amount_paid', 'hist_ex_latest_mean_plan_list_price', 
    'hist_ex_latest_mean_coverage_days', 'hist_ex_latest_mean_discount_amount',
    'hist_latest_payment_plan_days', 'hist_latest_plan_list_price', 
    'hist_latest_actual_amount_paid', 'hist_latest_coverage_days'
]

# Apply Explicit Sentinels
for col, fill_val in explicit_sentinel_fills.items():
    if col in df.columns:
        df[col] = df[col].fillna(fill_val)

# Apply Absolute Zeros
for col in absolute_zero_fills:
    if col in df.columns:
        df[col] = df[col].fillna(0.0)

print("Imputations complete. Verifying transaction feature matrix null states...")

# Final validation loop
tx_check_list = list(mode_fallbacks.keys()) + list(explicit_sentinel_fills.keys()) + absolute_zero_fills
all_clear = True
for col in tx_check_list:
    if col in df.columns:
        null_count = df[col].isna().sum()
        if null_count > 0:
            print(f"⚠️ Warning: {col} still has {null_count} NaNs!")
            all_clear = False

if all_clear:
    print("🎉 Success! Every transaction feature has been fully resolved without a single flag conflict.")
mode_zero_fills = [
    'hist_ex_latest_mode_payment_plan_days',
    'hist_ex_latest_mode_plan_list_price',
    'hist_ex_latest_mode_actual_amount_paid'
]

print("Patching remaining zero-transaction historical mode gaps...")
for col in mode_zero_fills:
    if col in df.columns:
        df[col] = df[col].fillna(0.0)
        print(f"-> {col} remaining NaNs: {df[col].isna().sum()}")

print("\n🎉 Verification check complete. Every column is now 100% clean!")    
```
