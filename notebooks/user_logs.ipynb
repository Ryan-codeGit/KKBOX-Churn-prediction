```python
df_clean = df[df['total_secs_w0'] != 'total_secs_w0']
```

```python
df_clean.shape
```

```python
df_clean['total_secs_w0'] = df_clean['total_secs_w0'].astype(str).str.strip()
df_clean['total_secs_w0'] = pd.to_numeric(df_clean['total_secs_w0'], errors='coerce')

# 4. Now find the true maximum number
print("The true mathematical maximum is:")
print(df_clean['total_secs_w0'].max())
```

```python
df_clean['hours_per_day'] = (df_clean['total_secs_w0'] / 7) / 3600

print("--- Distribution of Daily Listening Hours ---")
print(df_clean['hours_per_day'].describe(percentiles=[0.5, 0.9, 0.95, 0.99, 0.999]))
```
